```python
def analytical_energy(N, edge_full_data, num):
  analytical_energies = []  # analytical result

  for k in range(num):
    H = np.zeros((2**N, 2**N), dtype=complex)  # numpy 영행렬

    for (i, j), (J_xx, J_yy, J_zz) in edge_full_data[k].items():
      # print(q1, q2)
      # print(J_xx.item(), J_yy, J_zz)
      H += J_xx.item() * (
          qu.ikron(qu.pauli('X'), dims=[2]*N, inds=[i]) @
          qu.ikron(qu.pauli('X'), dims=[2]*N, inds=[j])
      )
      # YY 항
      H += J_yy.item() * (
          qu.ikron(qu.pauli('Y'), dims=[2]*N, inds=[i]) @
          qu.ikron(qu.pauli('Y'), dims=[2]*N, inds=[j])
      )
      # ZZ 항
      H += J_zz.item() * (
          qu.ikron(qu.pauli('Z'), dims=[2]*N, inds=[i]) @
          qu.ikron(qu.pauli('Z'), dims=[2]*N, inds=[j])
      )
    analytical_energies.append(qu.linalg.base_linalg.groundenergy(H))
  return analytical_energies

def mse(vector1, vector2):
    return np.mean((np.array(vector1) - np.array(vector2)) ** 2)
```

```python
def analytical_state(N, edge_full_data, num):
  analytical_states = []  # analytical result

  for k in range(num):
    H = np.zeros((2**N, 2**N), dtype=complex)  # numpy 영행렬

    for (i, j), (J_xx, J_yy, J_zz) in edge_full_data[k].items():
      # print(q1, q2)
      # print(J_xx.item(), J_yy, J_zz)
      H += J_xx.item() * (
          qu.ikron(qu.pauli('X'), dims=[2]*N, inds=[i]) @
          qu.ikron(qu.pauli('X'), dims=[2]*N, inds=[j])
      )
      # YY 항
      H += J_yy.item() * (
          qu.ikron(qu.pauli('Y'), dims=[2]*N, inds=[i]) @
          qu.ikron(qu.pauli('Y'), dims=[2]*N, inds=[j])
      )
      # ZZ 항
      H += J_zz.item() * (
          qu.ikron(qu.pauli('Z'), dims=[2]*N, inds=[i]) @
          qu.ikron(qu.pauli('Z'), dims=[2]*N, inds=[j])
      )
    analytical_states.append(qu.linalg.base_linalg.groundstate(H).reshape(-1))
  return analytical_states

```

```python
sum_of_energy = []
def train(n, d, train_edge_full_data, train_latent_data, num_data, latent_size, NN_shape, maxiter=1000, lr=0.001, stddev=0.1, dropout_rate=0.1):
    model = NN_MERA_Model( n, d, stddev, NN_shape, latent_size, dropout_rate)
    optimizer = torch.optim.Adam(model.parameters(), lr=lr)
    
    # decay_steps = 50
    # decay_rate = 0.7
    # scheduler = torch.optim.lr_scheduler.LambdaLR(
    #     optimizer, lr_lambda=lambda step: decay_rate ** (step // decay_steps)
    # )
    batch_size = 20
    lst = list(range(num_data))      
    
    for i in range(1, maxiter + 1):
        
        # for j in range(num_data):
        #     index = batch_size
        total_energy = 0.0
        for start in range(0, num_data, batch_size):
            chunk = lst[start : start + batch_size]   # 슬라이싱으로 5개 추출
            # print(chunk)
            batch_energy = torch.tensor(0.0, dtype=torch.float32)
            optimizer.zero_grad()
            for j in chunk:
                # print(j)
                edge_data = [t.tolist() for t in train_edge_full_data[j].values()]
                edges = list(train_edge_full_data[j].keys())
                graph_latent = train_latent_data[j]
                # print(edges)
                # print(edge_data)
                # print(graph_latent)
                energy,_ = model(edges, edge_data, graph_latent)
                batch_energy += energy
                total_energy += energy.item()
            # total_energy = total_energy.to(torch.float32)
            batch_energy.backward()
            optimizer.step()
            # scheduler.step()
        sum_of_energy.append(total_energy)
        if i % 10 == 0:
            print(f"Epoch {i}, Total Energy: {total_energy}")
    
        torch.save(model.state_dict(), "NN_VQE_HEA_1layer.pth")
```

```python

def calculate_fidelity(psi, phi):
    # 정규화
    phi = phi.detach().numpy()
    # print(psi)
    # print(phi)
    psi = psi / np.linalg.norm(psi)
    phi = phi / np.linalg.norm(phi)
    
    # 전역 위상 정렬
    phase = np.vdot(psi, phi) / abs(np.vdot(psi, phi))
    phi_aligned = phi / phase
    
    # 퍼델리티 계산
    fidelity = abs(np.vdot(psi, phi_aligned))**2
    return fidelity
def fidelity_torch(psi, phi, eps=1e-12):
    # 동일 shape·dtype 체크 생략 가능
    # psi = psi.detach().numpy()
    # arr = np.asarray(q.full()).ravel()           # shape (d,), 복소 ndarray
    # return torch.as_tensor(arr, dtype=dtype, device=device)
    psi = torch.from_numpy(psi)
    # phi = phi.detach().numpy()
    # print(psi)
    # print(phi)
    psi = psi / torch.linalg.vector_norm(psi)  # 자동으로 복소·실수 모두 처리
    phi = phi / torch.linalg.vector_norm(phi)

    return torch.abs(torch.vdot(psi.conj(), phi))**2 

def calculate_dataset_fidelities(states_A, states_B):
    assert len(states_A) == len(states_B), "Datasets must have the same number of states."
    
    fidelities = [fidelity_torch(psi, phi) for psi, phi in zip(states_A, states_B)]

    return fidelities

```
